```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
import os

# find the original dataset: next to this notebook, or in the project's "1 - Data Set and processing" folder
for data_dir in ["./", "../1 - Data Set and processing/"]:
    if os.path.exists(data_dir + "dataset_small.csv"):
        break
else:
    raise FileNotFoundError("dataset_small.csv (the original dataset) was not found. Put it next to this notebook.")

df = pd.read_csv(data_dir + "dataset_small.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum().sum())

print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed successfully!")
print("Rows remaining:", df.shape[0])
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
print(df["phishing"].value_counts())
```

```python
print(df[["phishing"]].head(10))
```

```python
df.to_csv("phishing_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final dataset shape:", df.shape)
```

```python
from sklearn.model_selection import train_test_split

print("Train-test split tools are ready!")
```

```python
X = df.drop(columns=["phishing"])
y = df["phishing"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
X_train.assign(phishing=y_train).to_csv(
    "phishing_train.csv", index=False
)

X_test.assign(phishing=y_test).to_csv(
    "phishing_test.csv", index=False
)

print("Training and testing files saved successfully!")
```

```python
print("Training dataset shape:", X_train.shape)
print("Testing dataset shape:", X_test.shape)

print("\nTraining label distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting label distribution:")
print(y_test.value_counts(normalize=True).round(3))
```
